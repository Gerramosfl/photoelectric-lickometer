```python
#SELECCIONADO
import pandas as pd
import matplotlib.pyplot as plt

def process_data(filepath):
    data = pd.read_csv(filepath)
    data.columns = data.columns.str.strip()  # Eliminar espacios en blanco en los nombres de las columnas

    # Calcular el incremento en 'LeftDuration'
    data['Duration Increase'] = data['LeftDuration'].diff().fillna(data['LeftDuration'])

    # Filtrar los datos para los cambios en 'LeftDuration'
    data = data[data['Duration Increase'] >= 0.5]

    # Calcular el tiempo de inicio de cada evento
    data['Event Start Time'] = data['ElapsedSecs'] - data['Duration Increase']

    return data[['Event Start Time', 'Duration Increase']]

# Crear la figura y el eje
fig, ax = plt.subplots(figsize=(20, 2))  

# Lista de archivos
filepaths = [BASE_DIR / "Magister_en_Neurobiología/Experimentación_básica_en_Neuro/Prueba_2/Agua_2/4/SIP004_010100_00.CSV"]

# Generar etiquetas automáticamente 
labels = [f'WT{i+5}' for i in range(len(filepaths))]  # Genera etiquetas

# Altura de cada barra
bar_height = 0.025

# Espacio entre barras
bar_spacing = 0.05

# Procesar y agregar los eventos al gráfico
for i, filepath in enumerate(filepaths):
    data = process_data(filepath)
    bars = [(start, max(duration, 1)) for start, duration in zip(data['Event Start Time'], data['Duration Increase'])]
    y_pos = i * (bar_height + bar_spacing)  
    ax.broken_barh(bars, (y_pos, bar_height), facecolors='black', label=labels[i])

# Configuración de los ejes
ax.set_xlim(2675, 2820)
ax.set_ylim(0, len(filepaths) * (bar_height + bar_spacing))
ax.set_xlabel('Tiempo (s)')
ax.set_yticks([i * (bar_height + bar_spacing) + bar_height / 2 for i in range(len(filepaths))])
ax.set_yticklabels(labels)
ax.tick_params(axis='x', rotation=45)
ax.tick_params(axis='both', which='major', labelsize=10)

ax.set_title('Microestructura de Consumo de Agua')

plt.show()
```

```python
# PARTON DE CONSUMO MODIFICADO AUTOMATIZADO PARA TODOS LOS DIAS Y TODAS LAS SEMANAS, 3 grupos 2, 6 y 9 meses
# ===========================
# Automatizador fiel + carpeta elegida por el usuario
# ===========================
import os, re
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# --- CONFIGURA AQUÍ TU CARPETA BASE ---
BASE_DIR = BASE_DIR / "proyecto_A"  # <- cámbiala a gusto
OUTPUT_DIR = Path(BASE_DIR) / "patrones_de_consumo"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
print(f"[DIR] Guardaré las figuras en: {OUTPUT_DIR}")

# ---------- Si ya tienes process_data en otra celda, puedes omitir esta definición ----------
def process_data(filepath):
    # Tu función original (CSV crudo)
    df = pd.read_csv(filepath, encoding="ISO-8859-1")
    df.columns = df.columns.str.strip()
    df['Duration Increase'] = df['LeftDuration'].diff().fillna(df['LeftDuration'])
    df = df[df['Duration Increase'] >= 0.5]
    df['Event Start Time'] = df['ElapsedSecs'] - df['Duration Increase']
    return df[['Event Start Time', 'Duration Increase']]

def did_time_por_dia(dia:int) -> int:
    """Días 1–3: 2h (7200s); Día 4: 4h (14400s)."""
    return 14400 if dia == 4 else 7200

def get_events_from_file(path: str) -> pd.DataFrame:
    """
    Devuelve DataFrame con:
      - 'Event Start Time'
      - 'Duration Increase'
    Si es CSV -> usa process_data().
    Si es *_procesado.xlsx -> lee columnas ya procesadas.
    """
    if not path:
        return pd.DataFrame(columns=['Event Start Time','Duration Increase'])

    if path.lower().endswith(".csv"):
        try:
            return process_data(path)
        except Exception as e:
            print(f"[WARN] CSV no legible {path}: {e}")
            return pd.DataFrame(columns=['Event Start Time','Duration Increase'])

    if path.lower().endswith(".xlsx"):
        try:
            df = pd.read_excel(path)
        except Exception as e:
            print(f"[WARN] XLSX no legible {path}: {e}")
            return pd.DataFrame(columns=['Event Start Time','Duration Increase'])
        df.columns = [str(c).strip() for c in df.columns]
        needed = {'Event Start Time','Duration Increase'}
        if not needed.issubset(df.columns):
            print(f"[WARN] Faltan columnas en {os.path.basename(path)} -> {sorted(df.columns)[:6]}")
            return pd.DataFrame(columns=['Event Start Time','Duration Increase'])
        out = df[['Event Start Time','Duration Increase']].copy()
        out['Event Start Time'] = pd.to_numeric(out['Event Start Time'], errors='coerce')
        out['Duration Increase'] = pd.to_numeric(out['Duration Increase'], errors='coerce')
        out = out.dropna(subset=['Event Start Time','Duration Increase'])
        out = out.sort_values('Event Start Time').reset_index(drop=True)
        return out

    print(f"[WARN] Extensión no soportada: {path}")
    return pd.DataFrame(columns=['Event Start Time','Duration Increase'])

def recolectar(globs: dict, pattern: str):
    """
    Toma variables globales tipo:
      dosmeses_etoh_semana{w}_dia{d}
      seismeses_etoh_semana{w}_dia{d}
      nuevemeses_etoh_semana{w}_dia{d}
    y devuelve dict[(semana, dia)] = [paths...]
    """
    rx = re.compile(pattern)
    out = {}
    for name, val in globs.items():
        m = rx.match(name)
        if not m:
            continue
        if not isinstance(val, (list, tuple)):
            continue
        try:
            w = int(m.group(1)); d = int(m.group(2))
        except Exception:
            continue
        out[(w,d)] = list(val)
    return out

def plot_raster_tres_bandas(filepaths_2m, filepaths_6m, filepaths_9m, semana:int, dia:int):
    """
    Raster con tres bandas (arriba->abajo):
      - 2 meses (verde)
      - 6 meses (rojo)
      - 9 meses (azul)
    Eje X en segundos (0–7200 o 0–14400 según el día).
    """
    bar_height = 0.12
    gap = 0.06  # separación vertical entre bandas
    y_top = 0.02
    y_2m = (y_top, bar_height)
    y_6m = (y_top + bar_height + gap, bar_height)
    y_9m = (y_top + 2*(bar_height + gap), bar_height)

    fig, ax = plt.subplots(figsize=(20, 2.8))

    # Banda 2 meses (verde)
    for fp in filepaths_2m:
        df = get_events_from_file(fp)
        if not df.empty:
            bars = [(float(s), max(float(d), 1.0)) for s, d in zip(df['Event Start Time'], df['Duration Increase'])]
            if bars:
                ax.broken_barh(bars, y_2m, facecolors='green')

    # Banda 6 meses (rojo)
    for fp in filepaths_6m:
        df = get_events_from_file(fp)
        if not df.empty:
            bars = [(float(s), max(float(d), 1.0)) for s, d in zip(df['Event Start Time'], df['Duration Increase'])]
            if bars:
                ax.broken_barh(bars, y_6m, facecolors='red')

    # Banda 9 meses (azul)
    for fp in filepaths_9m:
        df = get_events_from_file(fp)
        if not df.empty:
            bars = [(float(s), max(float(d), 1.0)) for s, d in zip(df['Event Start Time'], df['Duration Increase'])]
            if bars:
                ax.broken_barh(bars, y_9m, facecolors='blue')

    # Ejes y título
    xlim = did_time_por_dia(dia)
    ax.set_xlim(0, xlim)
    ax.set_ylim(0, y_top + 3*bar_height + 2*gap + 0.02)
    ax.set_xlabel('Tiempo (s)')
    # Posicionar etiquetas en el centro de cada banda
    yticks = [
        y_2m[0] + bar_height/2,
        y_6m[0] + bar_height/2,
        y_9m[0] + bar_height/2,
    ]
    ax.set_yticks(yticks)
    ax.set_yticklabels(['2 meses', '6 meses', '9 meses'])
    ax.tick_params(axis='x', rotation=45)
    ax.tick_params(axis='both', which='major', labelsize=10)
    ax.set_title(f'Patrón de consumo etanol – Semana {semana}, Día {dia}')

    plt.tight_layout()
    out = OUTPUT_DIR / f"raster_semana{semana}_dia{dia}.png"
    plt.savefig(out, dpi=200)
    plt.close(fig)
    print(f"[OK] Guardado: {out}")

# ---------- Ejecuta todo en lote (usa tus variables ya definidas arriba) ----------
_globs = globals()
# Mapeo por patrón de nombre de variables:
mapa_2m = recolectar(_globs, r"^dosmeses_etoh_semana(\d+)_dia(\d+)$")
mapa_6m = recolectar(_globs, r"^seismeses_etoh_semana(\d+)_dia(\d+)$")
mapa_9m = recolectar(_globs, r"^nuevemeses_etoh_semana(\d+)_dia(\d+)$")

generados = 0
for semana in range(1, 7):
    for dia in range(1, 5):
        filepaths_2m = mapa_2m.get((semana, dia), [])
        filepaths_6m = mapa_6m.get((semana, dia), [])
        filepaths_9m = mapa_9m.get((semana, dia), [])
        if filepaths_2m or filepaths_6m or filepaths_9m:
            plot_raster_tres_bandas(filepaths_2m, filepaths_6m, filepaths_9m, semana, dia)
            generados += 1

if generados == 0:
    print("[INFO] No se generaron gráficos (¿todas las listas están vacías?).")
else:
    print(f"[DONE] Se generaron {generados} figuras en {OUTPUT_DIR}")
```

```python
#PARTON DE CONSUMO MODIFICADO AUTOMATIZADO PARA TODOS LOS DIAS Y TODAS LAS SEMANAS 2 grupos, 6 meses y 9 meses

# ===========================
# Automatizador fiel + carpeta elegida por el usuario
# ===========================
import os, re
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# --- CONFIGURA AQUÍ TU CARPETA BASE ---
BASE_DIR = BASE_DIR / "proyecto_A/6xDID1"  # <- cámbiala a gusto
OUTPUT_DIR = Path(BASE_DIR) / "patrones_de_consumo"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
print(f"[DIR] Guardaré las figuras en: {OUTPUT_DIR}")

# ---------- Si ya tienes process_data en otra celda, puedes omitir esta definición ----------
def process_data(filepath):
    # Tu función original (CSV crudo)
    df = pd.read_csv(filepath, encoding="ISO-8859-1")
    df.columns = df.columns.str.strip()
    df['Duration Increase'] = df['LeftDuration'].diff().fillna(df['LeftDuration'])
    df = df[df['Duration Increase'] >= 0.5]
    df['Event Start Time'] = df['ElapsedSecs'] - df['Duration Increase']
    return df[['Event Start Time', 'Duration Increase']]

def did_time_por_dia(dia:int) -> int:
    """Días 1–3: 2h (7200s); Día 4: 4h (14400s)."""
    return 14400 if dia == 4 else 7200

def get_events_from_file(path: str) -> pd.DataFrame:
    """
    Devuelve DataFrame con:
      - 'Event Start Time'
      - 'Duration Increase'
    Si es CSV -> usa process_data().
    Si es *_procesado.xlsx -> lee columnas ya procesadas.
    """
    if not path:
        return pd.DataFrame(columns=['Event Start Time','Duration Increase'])

    if path.lower().endswith(".csv"):
        try:
            return process_data(path)
        except Exception as e:
            print(f"[WARN] CSV no legible {path}: {e}")
            return pd.DataFrame(columns=['Event Start Time','Duration Increase'])

    if path.lower().endswith(".xlsx"):
        try:
            df = pd.read_excel(path)
        except Exception as e:
            print(f"[WARN] XLSX no legible {path}: {e}")
            return pd.DataFrame(columns=['Event Start Time','Duration Increase'])
        df.columns = [str(c).strip() for c in df.columns]
        needed = {'Event Start Time','Duration Increase'}
        if not needed.issubset(df.columns):
            print(f"[WARN] Faltan columnas en {os.path.basename(path)} -> {sorted(df.columns)[:6]}")
            return pd.DataFrame(columns=['Event Start Time','Duration Increase'])
        out = df[['Event Start Time','Duration Increase']].copy()
        out['Event Start Time'] = pd.to_numeric(out['Event Start Time'], errors='coerce')
        out['Duration Increase'] = pd.to_numeric(out['Duration Increase'], errors='coerce')
        out = out.dropna(subset=['Event Start Time','Duration Increase'])
        out = out.sort_values('Event Start Time').reset_index(drop=True)
        return out

    print(f"[WARN] Extensión no soportada: {path}")
    return pd.DataFrame(columns=['Event Start Time','Duration Increase'])

def recolectar(globs: dict, pattern: str):
    """
    Toma variables globales tipo:
      seismeses_etoh_semana{w}_dia{d}
      nuevemeses_etoh_semana{w}_dia{d}
    y devuelve dict[(semana, dia)] = [paths...]
    """
    import re as _re
    rx = _re.compile(pattern)
    out = {}
    for name, val in globs.items():
        m = rx.match(name)
        if not m:
            continue
        if not isinstance(val, (list, tuple)):
            continue
        try:
            w = int(m.group(1)); d = int(m.group(2))
        except:
            continue
        out[(w,d)] = list(val)
    return out

def plot_raster_simple(filepaths_6m, filepaths_9m, semana:int, dia:int):
    """
    Gráfico original, con bandas corregidas:
      - Banda superior = 6 meses (rojo)
      - Banda inferior = 9 meses (azul)
      - broken_barh, eje en segundos
    """
    bar_height = 0.1
    fig, ax = plt.subplots(figsize=(20, 2.2))

    # Banda superior: 6 meses (rojo)
    for fp in filepaths_6m:
        df = get_events_from_file(fp)
        bars = [(float(s), max(float(d), 1.0)) for s, d in zip(df['Event Start Time'], df['Duration Increase'])]
        if bars:
            ax.broken_barh(bars, (0.02, bar_height), facecolors='red')

    # Banda inferior: 9 meses (azul)
    for fp in filepaths_9m:
        df = get_events_from_file(fp)
        bars = [(float(s), max(float(d), 1.0)) for s, d in zip(df['Event Start Time'], df['Duration Increase'])]
        if bars:
            ax.broken_barh(bars, (0.06 + bar_height, bar_height), facecolors='blue')

    # Ejes y título
    xlim = did_time_por_dia(dia)
    ax.set_xlim(0, xlim)
    ax.set_ylim(0, bar_height * 3)
    ax.set_xlabel('Tiempo (s)')
    ax.set_yticks([bar_height * 0.6, bar_height * 2])
    ax.set_yticklabels(['6 meses', '9 meses'])  # << orden corregido
    ax.tick_params(axis='x', rotation=45)
    ax.tick_params(axis='both', which='major', labelsize=10)
    ax.set_title(f'Patrón de consumo etanol – Semana {semana}, Día {dia}')

    plt.tight_layout()
    out = OUTPUT_DIR / f"raster_semana{semana}_dia{dia}.png"
    plt.savefig(out, dpi=200)
    plt.show()
    print(f"[OK] Guardado: {out}")

# ---------- Ejecuta todo en lote (usa tus variables ya definidas arriba) ----------
_globs = globals()
# Mapeo de variables a cada banda (corregido):
#   - banda superior (rojo): '6 meses' -> variables: seismeses_etoh_semanaX_diaY
#   - banda inferior (azul): '9 meses' -> variables: nuevemeses_etoh_semanaX_diaY
mapa_6m = recolectar(_globs, r"^seismeses_etoh_semana(\d+)_dia(\d+)$")
mapa_9m = recolectar(_globs, r"^nuevemeses_etoh_semana(\d+)_dia(\d+)$")

generados = 0
for semana in range(1, 7):
    for dia in range(1, 5):
        filepaths_6m = mapa_6m.get((semana, dia), [])  # arriba (rojo) -> 6 meses
        filepaths_9m = mapa_9m.get((semana, dia), [])  # abajo (azul) -> 9 meses
        if filepaths_6m or filepaths_9m:
            plot_raster_simple(filepaths_6m, filepaths_9m, semana, dia)
            generados += 1

if generados == 0:
    print("[INFO] No se generaron gráficos (¿todas las listas están vacías?).")
else:
    print(f"[DONE] Se generaron {generados} figuras en {OUTPUT_DIR}")
```

```python
# =========================================================
# Gráficos 3D separados por grupo (6m y 9m) con Z instantáneo (0/1)
#   - X: tiempo (min)
#   - Y: índice de sesión (S{semana}-D{día}) con offsets por animal
#   - Z: señal instantánea 0/1 (evento/no evento)
# =========================================================
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter

# Resolución de muestreo (segundos) para construir la señal 0/1
RES_SEC = 1.0

def _instantaneo_desde_eventos(df_events: pd.DataFrame, total_secs: int):
    """
    Devuelve:
      t (s) y occ (0/1) para cada instante muestreado.
    """
    n = int(total_secs // RES_SEC) + 1
    t = np.linspace(0, total_secs, n)
    occ = np.zeros_like(t, dtype=float)
    if df_events is None or df_events.empty:
        return t, occ
    for s, d in zip(df_events['Event Start Time'].values, df_events['Duration Increase'].values):
        s_idx = int(max(0, np.floor(s / RES_SEC)))
        e_idx = int(min(n, np.ceil((s + d) / RES_SEC)))
        if e_idx > s_idx:
            occ[s_idx:e_idx] = 1.0
    return t, occ


def plot_3d_instantaneo_por_grupo(grupo_nombre: str, mapa: dict, color: str,
                                  semanas=range(1,7), dias=range(1,5),
                                  animal_gap=0.12, line_alpha=0.9, lw=1.6,
                                  z_amp=0.2):  # ← altura máxima de los eventos
    fig = plt.figure(figsize=(20, 20))
    ax = fig.add_subplot(111, projection='3d')

    session_idx = 0
    y_ticks, y_labels = [], []

    for semana in semanas:
        for dia in dias:
            paths = mapa.get((semana, dia), [])
            if not paths:
                continue

            total_secs = did_time_por_dia(dia)
            n_anim = len(paths)
            centers = [session_idx + (ai - (n_anim-1)/2)*animal_gap for ai in range(n_anim)]

            for fp, y_center in zip(paths, centers):
                df = get_events_from_file(fp)
                t, occ = _instantaneo_desde_eventos(df, total_secs)
                ax.plot(t/60.0, np.full_like(t, y_center), occ * z_amp,  # ← escala Z
                        color=color, alpha=line_alpha, linewidth=lw)

            y_ticks.append(session_idx)
            y_labels.append(f"S{semana}-D{dia}")
            session_idx += 1

    ax.set_xlabel("Tiempo (min)")
    ax.set_ylabel("Sesiones (Semana–Día)")
    ax.set_zlabel("Consumo instantáneo (0–1)")
    ax.set_yticks(y_ticks)
    ax.set_yticklabels(y_labels)

    # Limita Z a [0, z_amp] y formatea ticks para que sigan mostrando 0–1
    ax.set_zlim(0, 1)                 # eje de 0 a 1
    ax.set_zticks([0, 1])  
    ax.zaxis.set_major_formatter(FuncFormatter(lambda v, pos: f"{v / z_amp:.1f}"))

    ax.view_init(elev=25, azim=-60)
    plt.tight_layout()
    out = OUTPUT_DIR / f"consumo3D_instantaneo_{grupo_nombre.replace(' ', '').replace('ó','o')}.png"
    plt.savefig(out, dpi=300, bbox_inches='tight', pad_inches=0.3)
    plt.show()
    print(f"[OK] Figura 3D instantánea guardada en: {out}")



plot_3d_instantaneo_por_grupo("6 meses", mapa_6m, color="red",
                              semanas=range(1,7), dias=range(1,5),
                              z_amp=0.01, lw=1.0, line_alpha=0.8)

plot_3d_instantaneo_por_grupo("9 meses", mapa_9m, color="blue",
                              semanas=range(1,7), dias=range(1,5),
                              z_amp=0.01, lw=1.0, line_alpha=0.8)


```
